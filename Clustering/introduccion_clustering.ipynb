```python
from sklearn.datasets import make_blobs
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
x,y = make_blobs(n_samples= 100, centers=4, n_features=2, cluster_std=[1,1.5,2,2], random_state=7)

df= pd.DataFrame({
    "x1": x[:,0],
    "x2": x[:,1],
    "y": y
})

df.head()
```

```python
sns.scatterplot(x= "x1", y= "x2", hue= "y", data= df) 
#plt.scatter(x= df["x1"], y= df["x2"], c= df["y"] #Con Matplotlib
```

```python
kmeans= KMeans(n_clusters=4, random_state= 7)
y_pred= kmeans.fit_predict(x)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(8, 6))
ax[0].scatter(x= df["x1"], y= df["x2"], c= df["y"])
ax[0].set_title(f"Original {len(np.unique(y))} clusters")

ax[1].scatter(x= df["x1"], y= df["x2"], c= y_pred)
ax[1].set_title(f"Kmeans {len(np.unique(y_pred))} clusters")
```
