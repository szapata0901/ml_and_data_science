```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

os.chdir("C:/Users/SEBASTIAN/Documents/Cursos/curso_ciencia_de_datos/30_days_of_python/Introduccion_a_Machine_Learning/")
```

```python
data = pd.read_csv("diabetes.csv")
data.head(2)

data.dtypes
```

```python
print("Número original de datapoints\n", data.shape)

data= data.dropna()

print("Número actualizado de datapoints\n", data.shape)
```

```python
plt.rcParams["font.size"] =15 # Ajustar el tamaño de fuente
f= plt.figure(figsize=(8,4))

ax= f.add_subplot(1,1,1) # Crear un subplot o subtrama - al ser una sola figura es 1,1,1

data["Outcome"].hist(ax=ax, edgecolor= "black", linewidth= 2)
ax.set_xlim([-0.5, 1.5]) # Establece los límites en el eje x
ax.set_xticks([0,1]) # Establece la frecuencia de tick. Tenemos 0 y 1 que corresponden a Sí y No respectivamente
ax.set_xticklabels(["N", "Y"]) # Etiquetar xtick labels
ax.set_title("Diabetes ¿Y/N?")
ax.set_xlabel("Answer")
ax.set_ylabel("Count")
ax.set_ylim([0,510]) # Establece los límites superior/inferior del eje y
f.tight_layout() #Ajusta el tamaño de la figura
```

```python
#RETO 1
binvalues = [20, 25, 30, 35, 40, 85]

f= plt.figure(figsize=(8,4))
ax= f.add_subplot(1,1,1)

data["Age"].hist(ax=ax, edgecolor= "black", linewidth= 2, bins= binvalues)
ax.set_title("Age range of patiens")
ax.set_ylim([0,510])
ax.set_xlabel("Age")
ax.set_ylabel("Count")
f.tight_layout()
```

```python
#RETO 2

f= plt.figure(figsize=(8,4))
ax= f.add_subplot(1,1,1)

data["Age"].hist(ax=ax, edgecolor= "black", linewidth= 2, bins= 5)
ax.set_title("Age range of patiens")
ax.set_ylim([0,510])
ax.set_xlabel("Age")
ax.set_ylabel("Count")
f.tight_layout()
```

```python
plt.rcParams["font.size"]= 15
plt.figure(figsize=(8,4))

plt.scatter(data["Age"], data["BloodPressure"], alpha= 0.5)
plt.title("Blood pressure vs. age range of patients")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.tight_layout
```

```python
#Eliminar datos atipicos

data= data[data["BloodPressure"] > 0]
data.shape

plt.rcParams["font.size"]= 15
plt.figure(figsize=(8,4))

plt.scatter(data["Age"], data["BloodPressure"], alpha= 0.25)
plt.title("Blood pressure vs. age range of patients")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.tight_layout
```

```python
diabetic = data[data["Outcome"] == 1]
notdiabetic = data[data["Outcome"] == 0] 

#Personas con diabetes
plt.rcParams["font.size"] = 15
f= plt.figure(figsize=(8,8))
ax= f.add_subplot(2,1,1)

ax.scatter(diabetic["BloodPressure"], diabetic["Age"], alpha= 0.25, c= "r")
ax.set_title("Diabetic: Age range of patients v. Blood Pressure")
ax.set_ylabel("Age")
ax.set_ylabel("Blood Pressure")
f.tight_layout()

#Personas sin diabetes
plt.rcParams["font.size"] = 15
f= plt.figure(figsize=(8,8))
ax= f.add_subplot(2,1,1)

ax.scatter(notdiabetic["BloodPressure"], notdiabetic["Age"], alpha= 0.25, c= "b")
ax.set_title("Not diabetic: Age range of patients v. Blood Pressure")
ax.set_ylabel("Age")
ax.set_ylabel("Blood Pressure")
f.tight_layout()
```
