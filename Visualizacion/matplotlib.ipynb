```python
import matplotlib.pyplot as plt
import numpy as np

x= np.linspace(0,5,11)
y= x**2

fig= plt.figure()
axes= fig.add_axes([0.1,0.1,0.8,0.9])
axes2= fig.add_axes([0.2, 0.55, 0.4, 0.3]) #Puedo tener dos graficas en una sola imagen y ambas se pueden modificar de manera independientes

axes.plot(x,y, "b")
axes2.plot(y,x, "r")
axes2.set_facecolor('gray')
fig.show()
```

```python
#HACER MULTIPLES GRAFICAS

import matplotlib.pyplot as plt
import numpy as np

x= np.linspace(0,5,11)
y= np.sin(x)

#Forma 1
fig, axes= plt.subplots(1,2) #Dos graficas en 1 fila con 2 columnas
axes[0].plot(x,y, "red") # Grafica 1
axes[1].plot(y,x, "blue")# Grafica 2

#Forma 2
fig, (axes1, axes2)= plt.subplots(1,2)
axes1.plot(x,y, "red") #Grafica 1
axes2.plot(y,x, "blue") #Grafica 2

#Forma 3
fig, axes = plt.subplots(2,4)
axes[0,0].plot(x,y) #Se indica la posicion de la grafica con indices
axes[0,1].plot(x,np.sin(x)**2) #Se indica la posicion de la grafica con indices
axes[0,2].plot(x, np.tan(x)) #Se indica la posicion de la grafica con indices
axes[0,3].plot(x, np.cos(x)) #Se indica la posicion de la grafica con indices
axes[1,0].plot(y) #Se indica la posicion de la grafica con indices
axes[1,1].plot(x,np.sin(x)**3) #Se indica la posicion de la grafica con indices
axes[1,2].plot(x, np.tan(x)**3) #Se indica la posicion de la grafica con indices
axes[1,3].plot(x) #Se indica la posicion de la grafica con indices

fig.tight_layout() #Para que las graficas no se superpongan
```

```python
#PONER NOMBRE A LOS EJES Y TITULOS

import matplotlib.pyplot as plt
import numpy as np

x= np.linspace(0,5,11)
y= np.sin(x)

fig, axes= plt.subplots(1,2, figsize= (5,5)) # Con "figsize" puedo jugar con el tamaño de las figuras
axes[0].plot(x,y, label= "sin (x)") #Label es para establecer una etiqueta
axes[0].set_title("Relacion X y Y")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[0].legend(loc= "upper right", bbox_to_anchor=(1,1) ) #Poner una leyenda en la grafica y se puede establecer la posicion. 
#En la leyenda se puede elegir entre upper, lower, right, left y sus combinaciones. La funcion "bbox_to_anchor" sirve para cambiar tamaño y ubicacion de la leyenda

axes[1].plot(y,x)
axes[1].set_title("Relacion Y y X")
axes[1].set_xlabel("Y")
axes[1].set_ylabel("X")
```

```python
import matplotlib.pyplot as plt
import numpy as np

country= ["India", "Japan", "Mexico", "Colombia", "Germany"]
population= [1000,800,900,500,600]

plt.bar(country, population, width= 0.5, color=["red", "blue"])
plt.xticks(np.arange(len(country)), ("India", "Japon","Mexico", "Colombia", "Alemania"), rotation= 45) #Cambiar los nombres, rotation= girar los nombres
plt.show()

plt.barh(country, population) #Grafica de manera horizontal
plt.show()
```

```python
#GRAFICO DE CAJAS

import matplotlib.pyplot as plt
import numpy as np

data= np.random.randint(1,50,100) 

plt.boxplot(data, vert= False, patch_artist= True, notch= True) #vert= Girar, patch_artist= Rellenar la caja, notch= Enfocar los datos hacia la mediana
plt.show()

data= np.append(data, 200) #Agregar un dato, en este caso, uno atipico
plt.boxplot(data, vert= False, patch_artist= True, notch= True, showfliers= True) #showfliers= Mostrar o no datos atipicos
```
