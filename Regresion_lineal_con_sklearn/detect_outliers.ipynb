```python
def detect_outliers(data, column):
    Q3 = data[column].quantile(0.75)
    Q1 = data[column].quantile(0.25)
    IQR = Q3 - Q1
    upper_bound = Q3 + (1.5 * IQR)
    lower_bound = Q1 - (1.5 * IQR)
    
    outliers_sup = data[data[column] > upper_bound].index
    outliers_inf = data[data[column] < lower_bound].index
    
    outliers = list(outliers_sup.union(outliers_inf))
    
    return f'Hay {len(outliers)} valores atípicos en la variable {column}, y corresponden a los índices: {outliers}'
```
