```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import os

os.chdir("C:/Users/SEBASTIAN/Documents/Cursos/curso_ciencia_de_datos/30_days_of_python/estadistica_descriptiva/")
df= pd.read_csv("cars.csv")

df.head()
```

```python
df.corr()
sns.heatmap(df.corr(), cmap= "coolwarm")
```

```python
#MEDIDAS DE TENDENCIA CENTRAL

print(df["price_usd"].mean())
print(df["price_usd"].median())

df["price_usd"].plot.hist(bins=20)

print(df.groupby("engine_type").count())

audi_q7= df[(df["manufacturer_name"] == "Audi") & (df["model_name"]== "Q7")]
sns.displot(audi_q7, x= "price_usd", hue= "year_produced")
```

```python
#MEDIDAS DE DISPERSION

# Desviacion estandar
print(df["price_usd"].std())

#Rango
rango= df["price_usd"].max() - df["price_usd"].min()
print(rango)

#Cuartiles

Q1= df["price_usd"].quantile(q= 0.25) #Q1
mediana= df["price_usd"].median() #Q2
Q3= df["price_usd"].quantile(q= 0.75) #Q3
val_min= df["price_usd"].quantile(q= 0) #Valor maximo
val_max= df["price_usd"].quantile(q= 1) #Valor minimo

print(val_min, Q1, mediana, Q3, val_max)

iqr= Q3 - Q1
print(iqr)

min_limit= Q1 - 1.5*iqr
max_limit= Q1 + 1.5*iqr
print(min_limit, max_limit)

fig, axes= plt.subplots(1,2)
axes[0].hist(df["price_usd"], bins= 20)
axes[1].boxplot(df["price_usd"])
fig.tight_layout()
fig.show()

fig, ax_seaborn = plt.subplots() #Paso 1 
sns.boxplot(df, x= "engine_fuel", y= "price_usd", ax=ax_seaborn) #Paso 2. Al haber usado plt.subplots, obligatoriamente debo hacer el paso 1 y 2, de lo contrario, va a quedar la figura superpuesta
plt.show()
```
