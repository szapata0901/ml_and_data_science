```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Crear un DataFrame con datos de ejemplo
data = {
    'Nombre': ['Juan', 'María', 'Luis', 'Ana', 'Pedro'],
    'Edad': [25, 30, 21, 28, 35],
    'Puntaje': [70, 85, 90, 82, 88]
}
df = pd.DataFrame(data)

# Mostrar los primeros registros del DataFrame
print(df.head())
```

```python
plt.bar(df["Nombre"], df["Edad"])
plt.xlabel("\nNombre")
plt.ylabel("Edad\n")
plt.title("Personas de la casa\n")
plt.show()
```

```python
correlation= np.corrcoef(df["Edad"], df["Puntaje"])[0,1]
correlation= round(correlation,3)

if correlation < 0.8: #Ejemplo sencillo
    print(f"La correlación es débil y corresponde a {correlation}")
else:
    print(f"La correlación es fuerte y corresponde a {correlation}")

print("\n\nLa gráfica que muestra la correlación: \n")

plt.scatter(df["Edad"], df["Puntaje"])
plt.xlabel("Edad")
plt.ylabel("Puntaje")
plt.title("Edad en función del puntaje")

plt.show()
```
