```python
numeros_normales= np.random.normal(0,1,100)
numeros_normales

print(np.min(numeros_normales)) #Forma 1 de obtener el numero mas pequeño 
print(numeros_normales.min()) #Forma 1 de obtener el numero mas pequeño 

print(np.max(numeros_normales)) #Numero maximo
print(np.mean(numeros_normales)) #Promedio
print(np.std(numeros_normales)) #Desviacion estandar
print(np.median(numeros_normales)) #Mediana
```

```python
print(matriz)

print("Columna minima: ", np.amin(matriz, axis= 0)) #Valor minimo en la columna
print("Columna maxima: ", np.amax(matriz, axis= 0)) #Valor maximo en la columna

print("Fila minima: ", np.amin(matriz, axis= 1)) #Valor minimo en la fila
print("Fila maxima: ", np.amax(matriz, axis= 1)) #Valor maximo en la fila
```

```python

a = np.array([1,2,3])

print('Tile:   ', np.tile(a, 2)) # Repetir todo de "a" 2 veces
print('Repeat: ', np.repeat(a, 2)) # Repetir cada elemento de "a" 2 veces

b= np.tile(a,100)
c= np.repeat(a,100)

import pandas as pd

data= pd.DataFrame(
    {
       "Columna 1": b,
       "Columna 2": c 
    }
)

print(data)

plt.scatter(data["Columna 1"], data["Columna 2"], color= "grey")
plt.show()
```

```python
temperatura= np.arange(1,20,1)
presion= (a*2) + 1

print("Correlacion: " , np.corrcoef(temperatura, presion))

plt.plot(temperatura, presion)
plt.xlabel(" \nTemperatura (°C)")
plt.ylabel("Presion (atm) \n")
plt.title("Presion vs Temperatura \n")

plt.show()
```
