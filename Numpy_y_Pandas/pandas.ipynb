```python
df = {'Name': ['Sebastian', 'John', 'Claudia'],
        'Country':['Colombia', 'UK', 'Spain'], 
        'City': ['Santander de Q', 'London', 'Madrid']}
df= pd.DataFrame(df, columns= ["Name", "Country", "City"])

weight= [74, 78, 69]
df["Weight"] = weight  #Añadir una columna

heights = [173, 175, 169] 
df["Height"] = heights #Añadir una columna

print(df)
```

```python
df["Height"] = df["Height"] * 0.01
df
```

```python
def calcular_bmi():
    weights = df["Weight"]
    heights = df["Height"]
    bmi= []

    for w,h in zip(weights, heights):
        b= w/(h*h)
        bmi.append(b)
    return bmi

calcular_bmi()

df["BMI"] = bmi
df["BMI"] = round(df["BMI"],1)
df
```

```python
data= {
    "A": [1,2,3,4,5],
    "B": [6,7,8,9,10],
    "C": [11,12,13,14,15],
    "D": [16,17,18,19,20]
}

df = pd.DataFrame(data, index=["a","b","c","d","e"])

print("Matriz original\n")
print(df)

print("FUNCION LOC")

print(df.loc["a"]) #Accediendo a una fila por etiqueta
print(df.loc["b","B"]) # Accediendo a una fila y columna por etiqueta
print(df.loc[:,"A":"C"]) #Seleccionando columnas
print(df.loc["b":"d",:]) #Seleccionando filas

print("FUNCION ILOC")

print(df.iloc[0]) #Accediendo a una fila por ubicacion numerica
print(df.iloc[1,1]) # Accediendo a una fila y columna por ubicacion numerica
print(df.iloc[:,0:3]) #Seleccionando columnas
print(df.iloc[1:4,:]) #Seleccionando filas
```

```python
# Eliminar filas y columnas

data= {
    "A": [1,2,3,4,5],
    "B": [6,7,8,9,10],
    "C": [11,12,13,14,15],
    "D": [16,17,18,19,20]
}

df = pd.DataFrame(data, index=["a","b","c","d","e"])

print("Matriz original\n")
print(df)

print("\n")
print(df.drop("a", axis= 0)) #Eliminar filas
print(df.drop("A", axis= 1)) #Eliminar columnas

print(df.drop(df.index[0])) #Eliminar filas
print(df.drop(df.columns[0], axis=1)) #Eliminar columnas
```

```python
#TRABAJAR CON DATOS NULOS

data= {
    "Nombre": ["Jose","Juan","Camila", "Paola", np.nan],
    "Edad": [11,23,np.nan,12,30],
    "Peso": [None, 40,60,64,70]
}

df= pd.DataFrame(data)

print("Matriz original \n")
print(df)

print("\n")

print(df.isnull()) #Preguntar por los datos nulos. Cada que encuentre un "NaN" o "None" va a reportarlo como verdadero
print(df.isnull()*1) #Llevarlo a forma numerica. Los datos nulos los va a representar con el numero 1
print(df.fillna("Missing")) #Llenar los datos nulos y poner "Missing"
print(df.fillna(df.mean())) #Llenar los datos nulos con una condicion, en este caso, el promedio
print(df.interpolate())  #Hace una interpolacion de una serie asumiendo el valor que deberia seguir y llena el dato faltante
print(df.dropna()) #Elimina los datos faltantes
```

```python
# ESTADISTICAS

df= pd.read_csv("bestsellers.csv", sep= ",")

print(df.describe()) #Descripcion general de los datos
print(df['Author'].value_counts()) #Contar los valores que tengo de acuerdo a una o varias variables. Ej de acuerdo al autor
print(df["Genre"].value_counts()) #Contar los valores que tengo de acuerdo a una o varias variables. Ej de acuerdo al genero

print(df.sort_values('Year')) #Ordenar los valores de acuerdo a una variable. Ordena de mayor a menor
print(df.sort_values('Year', ascending=False)) #Ordenar los valores de acuerdo a una variable. Ordena de menor a mayor 

print(df.sort_values("Reviews"))
print(df.sort_values("Reviews", ascending= False)) 
```

```python
# AGRUPAMIENTO (GROUPBY)

df= pd.read_csv("bestsellers.csv", sep= ",")

print(df.groupby("Genre").count())  #Agrupar por una caracteristica y reportar el estadistico
print(df.groupby("Genre").mean()) 
print(df.groupby("Genre").std())
print(df.groupby("Genre").min())
print(df.groupby("Genre").max())
```

```python
# AGRUPAMIENTO AVANZADO (GROUPBY)

df= pd.read_csv("bestsellers.csv", sep= ",")

print(df.groupby("Author").mean().loc["William Davis"]) # Agrupar y conocer esos valores en la fila. Es decir, los datos de promedio del autor 'William Davis'
print(df.groupby("Author").agg(["mean","std"])) #Agrupar por varios estadisticos
print(df.groupby("Author").agg({"Price": ["mean", "std"], "Reviews": ["mean", "std"]})) #Agrupar por varias funciones y por varias columnas
print(df.groupby(["Author", "Price"]).count()) #Agrupa y calcula la media por autor y precio.
```
